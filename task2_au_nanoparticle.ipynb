```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/Au_nanoParticle_dataset.csv")
print(df.columns.tolist()) 


new_df = df[["N_total", "N_bulk", "N_surface", "R_avg"]]
```

```python
new_df.head(20)
```

```python
summary_stats = new_df.describe().T[["mean", "std", "25%", "50%", "75%"]]
summary_stats.columns = ["Mean", "Std Dev", "Q1 (25%)", "Median (Q2)", "Q3 (75%)"]
print(summary_stats)
```

```python
fig, axes = plt.subplots(1, 4, figsize=(20, 4))

for ax, col in zip(axes, new_df.columns):
    ax.hist(new_df[col], bins=30, color="steelblue", edgecolor="black")
    ax.set_title(col)
    ax.set_xlabel(col)
    ax.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
