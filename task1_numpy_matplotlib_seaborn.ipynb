```python
import matplotlib.pyplot as plt

exp_data = np.random.exponential(scale=1.0, size=100000)

plt.figure(figsize=(8, 5))
plt.hist(np.random.rand(100000), density=True, bins=100, histtype="step", color="blue", label="uniform")
plt.hist(np.random.randn(100000), density=True, bins=100, histtype="step", color="red", label="normal")
plt.hist(exp_data, density=True, bins=100, histtype="step", color="green", label="exponential")

plt.axis([-3, 5, 0, 1.1])
plt.legend(loc="upper right")
plt.title("Random distributions")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")
surf = ax.plot_surface(X, Y, Z, cmap="viridis", edgecolor="none")

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title("3D Surface Plot of Z = X² + Y²")
fig.colorbar(surf, shrink=0.5, aspect=10)
plt.show()
```
